```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

X = np.array([
    [20, 2000],
    [22, 2500],
    [25, 3000],
    [45, 8000],
    [50, 9000],
    [55, 10000]
])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = KMeans(n_clusters=2, random_state=42, n_init=10)

labels = model.fit_predict(X_scaled)

print("Customer Clusters:", labels)
print("Cluster Centers:")
print(scaler.inverse_transform(model.cluster_centers_))
```

```python
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

# 1. Create sample dataset
X = np.array([
    [1000], [1200], [1500], [1800],
    [2000], [2200], [2500], [2800],
    [3000], [3200], [3500], [3800]
])

y = np.array([
    20, 24, 30, 36, 40, 44,
    50, 56, 60, 64, 70, 76
])

# 2. Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# 3. Create model
model = LinearRegression()

# 4. Train model
model.fit(X_train, y_train)

# 5. Make predictions
y_pred = model.predict(X_test)

# 6. Evaluate model
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Machine Learning Workflow Completed!")
print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))
print("Mean Absolute Error:", mae)
print("R2 Score:", r2)

# 7. Predict for new data
new_data = np.array([[2700]])

prediction = model.predict(new_data)

print("\nPredicted House Price:", prediction[0])
```
